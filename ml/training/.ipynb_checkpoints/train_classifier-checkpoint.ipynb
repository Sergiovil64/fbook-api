```python
print("Antes:", len(df))

df = df.drop_duplicates(subset=["tweet_text"])

print("Después:", len(df))
```

```python
#LIMPIEZA DE TEXTO
import re
import html

def clean_text(text):

    text = html.unescape(str(text))

    # URLs
    text = re.sub(r"http\S+|www\S+", " ", text)

    # menciones
    text = re.sub(r"@\w+", "@user", text)

    # quitar solo el símbolo #
    text = re.sub(r"#", "", text)

    # espacios
    text = re.sub(r"\s+", " ", text)

    return text.strip()
```

```python
df["clean_text"] = df["tweet_text"].apply(clean_text)
```

```python
df["label"] = df["cyberbullying_type"].apply(
    lambda clase: 0
    if clase == "not_cyberbullying"
    else 1
)
```

```python
id2label = {
    0: "not_cyberbullying",
    1: "cyberbullying"
}

label2id = {
    "not_cyberbullying": 0,
    "cyberbullying": 1
}
```

```python
from sklearn.model_selection import train_test_split

train_df, temp_df = train_test_split(
    df,
    test_size=0.20,
    random_state=42,
    stratify=df["label"]
)

val_df, test_df = train_test_split(
    temp_df,
    test_size=0.50,
    random_state=42,
    stratify=temp_df["label"]
)
```

```python
print("Train:", len(train_df))
print("Validation:", len(val_df))
print("Test:", len(test_df))

print("\nTrain:")
print(train_df["label"].value_counts(normalize=True))

print("\nValidation:")
print(val_df["label"].value_counts(normalize=True))

print("\nTest:")
print(test_df["label"].value_counts(normalize=True))
```

```python
from datasets import Dataset

train_dataset = Dataset.from_pandas(train_df)

val_dataset = Dataset.from_pandas(val_df)

test_dataset = Dataset.from_pandas(test_df)
```

```python
from datasets import Dataset

train_dataset = Dataset.from_pandas(
    train_df[["clean_text", "label"]].reset_index(drop=True)
)

val_dataset = Dataset.from_pandas(
    val_df[["clean_text", "label"]].reset_index(drop=True)
)

test_dataset = Dataset.from_pandas(
    test_df[["clean_text", "label"]].reset_index(drop=True)
)
```

```python
#2 TRAIN
```

```python
from transformers import AutoTokenizer

tokenizer = AutoTokenizer.from_pretrained(
    "distilbert-base-uncased"
)
```

```python
def tokenize(batch):
    return tokenizer(
        batch["clean_text"],
        padding="max_length",
        truncation=True,
        max_length=128
    )
```

```python
train_dataset = train_dataset.map(tokenize, batched=True)
val_dataset = val_dataset.map(tokenize, batched=True)
test_dataset = test_dataset.map(tokenize, batched=True)
```

```python
train_dataset = train_dataset.remove_columns(["clean_text"])
val_dataset = val_dataset.remove_columns(["clean_text"])
test_dataset = test_dataset.remove_columns(["clean_text"])
```

```python
train_dataset = train_dataset.rename_column("label", "labels")
val_dataset = val_dataset.rename_column("label", "labels")
test_dataset = test_dataset.rename_column("label", "labels")
```

```python
train_dataset.set_format("torch")
val_dataset.set_format("torch")
test_dataset.set_format("torch")
```

```python
from transformers import AutoModelForSequenceClassification

binary_model = AutoModelForSequenceClassification.from_pretrained(
    "distilbert-base-uncased",
    num_labels=2,
    id2label=id2label,
    label2id=label2id
)
```

```python
import numpy as np

from sklearn.metrics import (
    accuracy_score,
    precision_recall_fscore_support
)

def compute_binary_metrics(eval_pred):
    logits, labels = eval_pred
    predictions = np.argmax(logits, axis=-1)

    accuracy = accuracy_score(labels, predictions)

    precision, recall, f1, _ = precision_recall_fscore_support(
        labels,
        predictions,
        average="binary",
        pos_label=1,
        zero_division=0
    )

    return {
        "accuracy": accuracy,
        "precision": precision,
        "recall": recall,
        "f1": f1
    }
```

```python
from transformers import TrainingArguments

training_args_binary = TrainingArguments(
    output_dir="./results_binary",

    num_train_epochs=5,

    per_device_train_batch_size=16,
    per_device_eval_batch_size=16,

    learning_rate=2e-5,
    weight_decay=0.01,
    warmup_ratio=0.1,

    eval_strategy="epoch",
    save_strategy="epoch",
    logging_strategy="steps",
    logging_steps=100,

    load_best_model_at_end=True,
    metric_for_best_model="f1",
    greater_is_better=True,

    save_total_limit=2,
    report_to="none",
    seed=42,

    fp16=torch.cuda.is_available()
)
```

```python
eval_strategy="epoch"
```

```python
#old
from transformers import Trainer, EarlyStoppingCallback

binary_trainer = Trainer(
    model=binary_model,
    args=training_args_binary,
    train_dataset=train_dataset,
    eval_dataset=val_dataset,
    compute_metrics=compute_binary_metrics,
    callbacks=[
        EarlyStoppingCallback(
            early_stopping_patience=2
        )
    ]
)
```

```python
binary_trainer.train()
```

```python
# 3 EVALUATE
```

```python
binary_test_results = binary_trainer.evaluate(test_dataset)

binary_test_results
```

```python
from sklearn.metrics import classification_report
from scipy.special import softmax

binary_predictions = binary_trainer.predict(test_dataset)

y_true = binary_predictions.label_ids

#probabilities = softmax(binary_predictions.predictions, axis=1)
#cyber_probs = probabilities[:, 1]
#threshold = 0.70
#y_pred = (cyber_probs >= threshold).astype(int)

y_pred = np.argmax(binary_predictions.predictions, axis=1)

print(
    classification_report(
        y_true,
        y_pred,
        target_names=[
            "not_cyberbullying",
            "cyberbullying"
        ],
        digits=4,
        zero_division=0
    )
)
```

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
import matplotlib.pyplot as plt

cm = confusion_matrix(y_true, y_pred)

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=[
        "Not cyberbullying",
        "Cyberbullying"
    ]
)

disp.plot(values_format="d")

plt.title("Binary Cyberbullying Confusion Matrix")
plt.show()
```

```python
# 4 PREDICT
```

```python
import torch

def predict_binary_cyberbullying(text):
    cleaned_text = clean_text(text)

    inputs = tokenizer(
        cleaned_text,
        return_tensors="pt",
        truncation=True,
        padding=True,
        max_length=128
    )

    device = next(binary_model.parameters()).device

    inputs = {
        key: value.to(device)
        for key, value in inputs.items()
    }

    binary_model.eval()

    with torch.no_grad():
        outputs = binary_model(**inputs)

    probabilities = torch.softmax(
        outputs.logits,
        dim=-1
    )[0]

    predicted_id = torch.argmax(probabilities).item()

    return {
        "text": text,
        "prediction": id2label[predicted_id],
        "confidence": round(
            probabilities[predicted_id].item() * 100,
            2
        )
    }
```

```python
print(
    predict_binary_cyberbullying(
        "You are too ugly."
    )
)

print(
    predict_binary_cyberbullying(
        "I hope everyone has a good day."
    )
)

print(
    predict_binary_cyberbullying(
        "you are stupid."
    )
)
```
